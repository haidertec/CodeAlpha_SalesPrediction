```python
import json
import logging
import sys
import time
import warnings
from dataclasses import dataclass
from pathlib import Path

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from scipy.stats import loguniform, randint, uniform
from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import (GradientBoostingRegressor, RandomForestRegressor,
                              StackingRegressor, VotingRegressor)
from sklearn.impute import SimpleImputer
from sklearn.inspection import permutation_importance
from sklearn.linear_model import (ElasticNet, Lasso, LinearRegression, Ridge,
                                  RidgeCV)
from sklearn.metrics import (mean_absolute_error, mean_absolute_percentage_error,
                             mean_squared_error, r2_score)
from sklearn.model_selection import (KFold, RandomizedSearchCV, cross_validate,
                                     learning_curve, train_test_split)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.tree import DecisionTreeRegressor

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid", context="notebook")
plt.rcParams["figure.dpi"] = 110
plt.rcParams["savefig.bbox"] = "tight"


@dataclass
class Config:
    random_state: int = 42
    test_size: float = 0.2
    cv_folds: int = 5
    n_jobs: int = -1
    n_iter_search: int = 25
    output_dir: Path = Path("artifacts_sales")
    fig_dpi: int = 110
    target_col: str = "Sales"
    data_file: Path = Path("Advertising.csv")


CFG = Config()
CFG.output_dir.mkdir(parents=True, exist_ok=True)


def setup_logger(name="sales_pred"):
    logger = logging.getLogger(name)
    if logger.handlers:
        return logger
    logger.setLevel(logging.INFO)
    fmt = logging.Formatter("[%(asctime)s] %(levelname)-8s | %(message)s",
                            datefmt="%H:%M:%S")
    sh = logging.StreamHandler(sys.stdout)
    sh.setFormatter(fmt)
    logger.addHandler(sh)
    return logger


LOG = setup_logger()


def banner(title, char="=", width=78):
    LOG.info(char * width)
    LOG.info(title.center(width))
    LOG.info(char * width)


class SalesDataLoader:
    def __init__(self, cfg):
        self.cfg = cfg

    def load(self):
        if not self.cfg.data_file.exists():
            LOG.warning("CSV not found - using synthetic fallback")
            return self._synthetic()
        df = pd.read_csv(self.cfg.data_file)
        if "Unnamed: 0" in df.columns:
            df = df.drop(columns=["Unnamed: 0"])
        LOG.info("Loaded -> %s | cols=%s", df.shape, list(df.columns))
        return df

    def _synthetic(self):
        rng = np.random.default_rng(self.cfg.random_state)
        n = 200
        tv = rng.uniform(0.7, 296, n)
        radio = rng.uniform(0, 50, n)
        news = rng.uniform(0, 115, n)
        sales = 2.9 + 0.046 * tv + 0.18 * radio + 0.002 * news + rng.normal(0, 1.2, n)
        return pd.DataFrame({"TV": tv.round(1), "Radio": radio.round(1),
                             "Newspaper": news.round(1), "Sales": sales.round(2)})


class SalesFeatureEngineer:
    def __init__(self, cfg):
        self.cfg = cfg
        self.feature_cols = []
        self.numeric_cols = []

    def transform(self, df):
        df = df.copy()
        df.columns = [c.strip() for c in df.columns]
        col_map = {c.lower(): c for c in df.columns}
        target = None
        for cand in ["sales", "sales_amount", "revenue"]:
            if cand in col_map:
                target = col_map[cand]
                break
        if target is None:
            raise ValueError("Sales column missing")
        if target != self.cfg.target_col:
            df = df.rename(columns={target: self.cfg.target_col})

        spend_cols = [c for c in df.columns
                      if c.lower() in {"tv", "radio", "newspaper",
                                       "social_media", "influencer",
                                       "advertising", "ad_spend"}]
        if spend_cols:
            df["total_spend"] = df[spend_cols].sum(axis=1)
            if len(spend_cols) > 1:
                for c in spend_cols:
                    df[f"{c}_share"] = df[c] / (df["total_spend"] + 1e-6)
            for c in spend_cols:
                df[f"{c}_sqrt"] = np.sqrt(df[c].clip(lower=0))
                df[f"{c}_log"] = np.log1p(df[c].clip(lower=0))
            if "TV" in df.columns and "Radio" in df.columns:
                df["tv_radio_interaction"] = df["TV"] * df["Radio"]

        df = df.dropna(subset=[self.cfg.target_col])
        self.feature_cols = [c for c in df.columns if c != self.cfg.target_col]
        self.numeric_cols = [c for c in self.feature_cols
                             if pd.api.types.is_numeric_dtype(df[c])]
        LOG.info("Features -> total=%d", len(self.feature_cols))
        return df

    def build_preprocessor(self):
        num_pipe = Pipeline([("imp", SimpleImputer(strategy="median")),
                             ("sc", StandardScaler())])
        return ColumnTransformer([("num", num_pipe, self.numeric_cols)],
                                 remainder="drop")


class SalesModelFactory:
    @staticmethod
    def build(pre):
        return {
            "LinearRegression": Pipeline([("pre", pre),
                                          ("reg", LinearRegression())]),
            "Ridge": Pipeline([("pre", clone(pre)),
                               ("reg", Ridge(random_state=CFG.random_state))]),
            "Lasso": Pipeline([("pre", clone(pre)),
                               ("reg", Lasso(random_state=CFG.random_state))]),
            "ElasticNet": Pipeline([("pre", clone(pre)),
                                    ("reg", ElasticNet(
                                        random_state=CFG.random_state))]),
            "PolynomialRidge": Pipeline([
                ("pre", clone(pre)),
                ("poly", PolynomialFeatures(degree=2, include_bias=False)),
                ("reg", Ridge(random_state=CFG.random_state))]),
            "DecisionTree": Pipeline([("pre", clone(pre)), ("reg",
                DecisionTreeRegressor(random_state=CFG.random_state))]),
            "RandomForest": Pipeline([("pre", clone(pre)), ("reg",
                RandomForestRegressor(random_state=CFG.random_state,
                                      n_jobs=CFG.n_jobs))]),
            "GradientBoosting": Pipeline([("pre", clone(pre)), ("reg",
                GradientBoostingRegressor(random_state=CFG.random_state))]),
        }

    @staticmethod
    def search_spaces():
        return {
            "LinearRegression": {},
            "Ridge": {"reg__alpha": loguniform(1e-3, 1e3)},
            "Lasso": {"reg__alpha": loguniform(1e-3, 1e2)},
            "ElasticNet": {"reg__alpha": loguniform(1e-3, 1e2),
                           "reg__l1_ratio": uniform(0.1, 0.8)},
            "PolynomialRidge": {"reg__alpha": loguniform(1e-3, 1e3)},
            "DecisionTree": {"reg__max_depth": randint(2, 20),
                             "reg__min_samples_split": randint(2, 15)},
            "RandomForest": {"reg__n_estimators": randint(100, 400),
                             "reg__max_depth": randint(3, 20)},
            "GradientBoosting": {"reg__n_estimators": randint(50, 300),
                                 "reg__learning_rate": loguniform(1e-2, 3e-1)},
        }


class SalesTrainer:
    def __init__(self, cfg):
        self.cfg = cfg
        self.cv = KFold(n_splits=cfg.cv_folds, shuffle=True,
                        random_state=cfg.random_state)
        self.best_models = {}
        self.results = {}

    def tune(self, name, pipe, X, y):
        space = SalesModelFactory.search_spaces()[name]
        if not space:
            pipe.fit(X, y)
            return pipe
        search = RandomizedSearchCV(pipe, space, n_iter=self.cfg.n_iter_search,
                                    scoring="neg_root_mean_squared_error",
                                    cv=self.cv, n_jobs=self.cfg.n_jobs,
                                    random_state=self.cfg.random_state,
                                    refit=True)
        t0 = time.perf_counter()
        search.fit(X, y)
        LOG.info("  |- %-18s tuned %.2fs | CV RMSE=%.4f", name,
                 time.perf_counter() - t0, -search.best_score_)
        return search.best_estimator_

    def evaluate(self, name, model, X_tr, y_tr, X_te, y_te):
        y_pred = model.predict(X_te)
        cv_res = cross_validate(model, X_tr, y_tr, cv=self.cv,
                                scoring=["r2", "neg_root_mean_squared_error"],
                                n_jobs=self.cfg.n_jobs)
        return {
            "test_rmse": float(np.sqrt(mean_squared_error(y_te, y_pred))),
            "test_mae": float(mean_absolute_error(y_te, y_pred)),
            "test_mape": float(mean_absolute_percentage_error(y_te, y_pred)),
            "test_r2": float(r2_score(y_te, y_pred)),
            "cv_r2_mean": float(cv_res["test_r2"].mean()),
            "y_pred": y_pred,
        }

    def build_ensembles(self, X_tr, y_tr):
        estimators = [(n, m) for n, m in self.best_models.items()]
        voting = VotingRegressor(estimators=estimators, n_jobs=self.cfg.n_jobs)
        voting.fit(X_tr, y_tr)
        stacking = StackingRegressor(
            estimators=estimators,
            final_estimator=RidgeCV(),
            cv=self.cv, n_jobs=self.cfg.n_jobs)
        stacking.fit(X_tr, y_tr)
        return {"Voting": voting, "Stacking": stacking}


class SalesVisualizer:
    def __init__(self, cfg):
        self.cfg = cfg
        self.out = cfg.output_dir

    def _save(self, fig, name):
        fig.savefig(self.out / name, dpi=self.cfg.fig_dpi, bbox_inches="tight")
        LOG.info("  |- %s", name)

    def eda(self, df, feature_cols):
        banner("EDA")
        num = df[feature_cols + [self.cfg.target_col]].select_dtypes(
            include=[np.number])

        fig, ax = plt.subplots(figsize=(11, 8))
        sns.heatmap(num.corr(), annot=True, fmt=".2f", cmap="coolwarm",
                    linewidths=0.5, ax=ax)
        ax.set_title("Correlation Matrix")
        plt.tight_layout()
        self._save(fig, "01_correlation.png")
        plt.show()

        cols = [c for c in feature_cols if c in df.columns][:6]
        if cols:
            fig, axes = plt.subplots(2, 3, figsize=(16, 9))
            for a, col in zip(axes.ravel(), cols):
                sns.regplot(data=df, x=col, y=self.cfg.target_col,
                            scatter_kws={"alpha": 0.5},
                            line_kws={"color": "red"}, ax=a)
                a.set_title(f"Sales vs {col}")
            plt.tight_layout()
            self._save(fig, "02_spend_vs_sales.png")
            plt.show()

        fig, ax = plt.subplots(figsize=(10, 6))
        sns.histplot(df[self.cfg.target_col], kde=True, bins=35,
                     color="#4C72B0", ax=ax)
        ax.set_title("Sales Distribution")
        plt.tight_layout()
        self._save(fig, "03_sales_distribution.png")
        plt.show()

    def predictions(self, y_te, y_pred, tag="Stacking"):
        fig, axes = plt.subplots(1, 2, figsize=(14, 6))
        axes[0].scatter(y_te, y_pred, alpha=0.6, color="#4C72B0")
        lim = [min(y_te.min(), y_pred.min()), max(y_te.max(), y_pred.max())]
        axes[0].plot(lim, lim, "r--", lw=2)
        axes[0].set(xlabel="Actual Sales", ylabel="Predicted Sales",
                    title=f"Actual vs Predicted - {tag}")
        axes[0].grid(alpha=0.3)
        residuals = y_te - y_pred
        axes[1].scatter(y_pred, residuals, alpha=0.6, color="#DD8452")
        axes[1].axhline(0, color="black", ls="--")
        axes[1].set(xlabel="Predicted", ylabel="Residuals",
                    title=f"Residuals - {tag}")
        axes[1].grid(alpha=0.3)
        plt.tight_layout()
        self._save(fig, "04_predictions.png")
        plt.show()

    def model_comparison(self, results):
        names = list(results.keys())
        r2 = [results[n]["test_r2"] for n in names]
        rmse = [results[n]["test_rmse"] for n in names]
        x = np.arange(len(names))
        fig, ax1 = plt.subplots(figsize=(13, 6))
        ax1.bar(x - 0.2, r2, 0.4, color="#4C72B0", label="R2")
        ax1.set_ylabel("R2", color="#4C72B0")
        ax1.set_ylim(0, 1.05)
        ax2 = ax1.twinx()
        ax2.bar(x + 0.2, rmse, 0.4, color="#C44E52", label="RMSE")
        ax2.set_ylabel("RMSE", color="#C44E52")
        ax1.set_xticks(x)
        ax1.set_xticklabels(names, rotation=25, ha="right")
        ax1.set_title("Model Comparison - R2 vs RMSE")
        ax1.grid(axis="y", alpha=0.3)
        plt.tight_layout()
        self._save(fig, "05_model_comparison.png")
        plt.show()

    def feature_importance(self, model, X_te, y_te):
        banner("PERMUTATION IMPORTANCE")
        r = permutation_importance(model, X_te, y_te, n_repeats=10,
                                   random_state=self.cfg.random_state,
                                   n_jobs=self.cfg.n_jobs)
        imp = pd.DataFrame({"feature": X_te.columns,
                            "importance": r.importances_mean}
                           ).sort_values("importance", ascending=False).head(15)
        fig, ax = plt.subplots(figsize=(10, 8))
        ax.barh(imp["feature"][::-1], imp["importance"][::-1], color="#55A868")
        ax.set(xlabel="Importance", title="Top-15 Feature Importance")
        ax.grid(axis="x", alpha=0.3)
        plt.tight_layout()
        self._save(fig, "06_feature_importance.png")
        plt.show()

    def learning_curve(self, model, X, y):
        sizes, tr_sc, va_sc = learning_curve(
            model, X, y, cv=self.cfg.cv_folds, scoring="r2",
            train_sizes=np.linspace(0.3, 1.0, 5),
            n_jobs=self.cfg.n_jobs, random_state=self.cfg.random_state)
        fig, ax = plt.subplots(figsize=(9, 6))
        ax.plot(sizes, tr_sc.mean(1), "o-", label="Training R2")
        ax.fill_between(sizes, tr_sc.mean(1) - tr_sc.std(1),
                        tr_sc.mean(1) + tr_sc.std(1), alpha=0.15)
        ax.plot(sizes, va_sc.mean(1), "s-", label="Validation R2")
        ax.fill_between(sizes, va_sc.mean(1) - va_sc.std(1),
                        va_sc.mean(1) + va_sc.std(1), alpha=0.15)
        ax.set(xlabel="Training samples", ylabel="R2", title="Learning Curve")
        ax.legend()
        ax.grid(alpha=0.3)
        plt.tight_layout()
        self._save(fig, "07_learning_curve.png")
        plt.show()


class SalesPipeline:
    def __init__(self, cfg):
        self.cfg = cfg
        self.loader = SalesDataLoader(cfg)
        self.engineer = SalesFeatureEngineer(cfg)
        self.trainer = SalesTrainer(cfg)
        self.viz = SalesVisualizer(cfg)

    def run(self):
        banner("SALES PREDICTION - CodeAlpha Task 4", char="#")
        t_start = time.perf_counter()

        df = self.engineer.transform(self.loader.load())
        self.viz.eda(df, self.engineer.feature_cols)

        X = df[self.engineer.feature_cols]
        y = df[self.cfg.target_col].values
        X_tr, X_te, y_tr, y_te = train_test_split(
            X, y, test_size=self.cfg.test_size,
            random_state=self.cfg.random_state)

        pre = self.engineer.build_preprocessor()
        pipelines = SalesModelFactory.build(pre)

        banner("HYPERPARAMETER TUNING")
        for name, pipe in pipelines.items():
            best = self.trainer.tune(name, pipe, X_tr, y_tr)
            self.trainer.best_models[name] = best
            self.trainer.results[name] = self.trainer.evaluate(
                name, best, X_tr, y_tr, X_te, y_te)
            LOG.info("  |   R2=%.4f | RMSE=%.4f | MAE=%.4f",
                     self.trainer.results[name]["test_r2"],
                     self.trainer.results[name]["test_rmse"],
                     self.trainer.results[name]["test_mae"])

        ensembles = self.trainer.build_ensembles(X_tr, y_tr)
        ens_results = {}
        for ename, emodel in ensembles.items():
            ens_results[ename] = self.trainer.evaluate(
                ename, emodel, X_tr, y_tr, X_te, y_te)
            LOG.info("  |- %-10s R2=%.4f | RMSE=%.4f", ename,
                     ens_results[ename]["test_r2"],
                     ens_results[ename]["test_rmse"])

        all_r2 = {**{k: v["test_r2"] for k, v in self.trainer.results.items()},
                  **{k: v["test_r2"] for k, v in ens_results.items()}}
        winner = max(all_r2, key=all_r2.get)
        LOG.info("* WINNER -> %s (R2=%.4f)", winner, all_r2[winner])
        winner_model = ensembles.get(winner) or self.trainer.best_models.get(winner)

        self.viz.model_comparison(self.trainer.results)
        winner_ens = ens_results.get("Stacking", ens_results["Voting"])
        self.viz.predictions(y_te, winner_ens["y_pred"], "Stacking/Voting")
        self.viz.feature_importance(winner_model, X_te, y_te)
        self.viz.learning_curve(winner_model, X_tr, y_tr)

        joblib.dump(winner_model, self.cfg.output_dir / f"best_model_{winner}.joblib")

        banner("INFERENCE DEMO ON NEW SAMPLES")
        self._demo(winner_model)

        banner(f"TASK 4 COMPLETE in {time.perf_counter() - t_start:.2f}s", char="#")

    def _demo(self, model):
        sample = pd.DataFrame({
            "TV": [150.0, 50.0, 250.0],
            "Radio": [25.0, 10.0, 40.0],
            "Newspaper": [30.0, 15.0, 60.0],
            "total_spend": [205.0, 75.0, 350.0],
            "TV_share": [0.73, 0.67, 0.71],
            "Radio_share": [0.12, 0.13, 0.11],
            "Newspaper_share": [0.15, 0.20, 0.17],
            "TV_sqrt": [12.25, 7.07, 15.81],
            "Radio_sqrt": [5.0, 3.16, 6.32],
            "Newspaper_sqrt": [5.48, 3.87, 7.75],
            "TV_log": [5.02, 3.93, 5.53],
            "Radio_log": [3.26, 2.40, 3.71],
            "Newspaper_log": [3.43, 2.77, 4.11],
            "tv_radio_interaction": [3750.0, 500.0, 10000.0],
        })
        for c in self.engineer.feature_cols:
            if c not in sample.columns:
                sample[c] = 0
        sample = sample[self.engineer.feature_cols]
        preds = model.predict(sample)
        for i, p in enumerate(preds, 1):
            LOG.info("  Sample %d -> Predicted Sales = %.2f units", i, p)


SalesPipeline(CFG).run()
```
